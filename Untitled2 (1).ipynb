### Cleaning the `County` Column

The `County` column contained inconsistent values, including the `"County"` suffix, leading/trailing whitespace, and `"Unknown"` entries.

To standardize the data:

- Removed the `"County"` suffix from county names.
- Trimmed unnecessary whitespace.
- Replaced `"Unknown"` values with `NaN` so they can be treated as missing data during analysis.

```python
df["County"] = (
    df["County"]
        .str.replace("County", "", regex=False)
        .str.strip()
        .replace("Unknown", np.nan)
)
```

### Filling Missing City Values

Some records had missing values in the `City` column while the corresponding `County` information was available. Since all counties in the dataset belong to New York State, missing city values were filled with `"New York"` based on the county information.

```python
import numpy as np

# Gönderdiğin liste (NaN değeri numpy ile ifade edilebilir veya yoksayılabilir, biz listede temizliyoruz)
county_list = [
    'Kings', 'Monroe', 'Nassau', 'Schenectady', 'Dutchess', 'Ulster', 'Greene', 
    'Oneida', 'Yates', 'Delaware', 'Suffolk', 'New York', 'Orleans', 'Niagara', 
    'Rensselaer', 'Clinton', 'Cattaraugus', 'Lewis', 'Genesee', 'Orange', 
    'Warren', 'Steuben', 'Broome', 'Erie', 'Schoharie', 'Columbia', 'Richmond', 
    'Schuyler', 'Jefferson', 'Putnam', 'Wayne', 'Tompkins', 'Chemung', 
    'Livingston', 'Montgomery', 'Ontario', 'Westchester', 'Bronx', 'St. Lawrence', 
    'Fulton', 'Washington', 'Essex', 'Madison', 'Tioga', 'Herkimer', 'Cortland', 
    'Otsego', 'Rockland', 'Chautauqua', 'Saratoga', 'Wyoming', 'Seneca', 'Allegany', 
    'Franklin', 'Onondaga', 'Sullivan', 'Cayuga', 'Albany', 'Hamilton', 'Oswego', 
    'Queens', 'Chenango'
]

# 1. Hepsini 'New York' iline eşleyen sözlüğü otomatik oluşturuyoruz
# Not: Sözlük { 'Kings': 'New York', 'Monroe': 'New York', ... } şeklinde oluşacak.
ilce_il_haritasi = {x: 'New York' for x in county_list}

# 2. 'city' sütunundaki boş (NaN) değerleri, 'county' sütununun haritadaki karşılığıyla dolduruyoruz
df['City'] = df['City'].fillna(df['County'].map(ilce_il_haritasi))
```

### Converting Enrollment Counts to Numeric Values

Several enrollment columns contained the value `"<4"` instead of a numeric count. Since these values cannot be analyzed numerically, they were replaced before converting the columns to numeric data types.

- In the **Total Attending SUNY Institutions** column, `"<4"` was replaced with **4** to preserve the minimum reported total.
- In the remaining enrollment columns, `"<4"` was replaced with **2** as an estimated midpoint value, allowing quantitative analysis while minimizing the impact of suppressed counts.

```python
df["Total Attending SUNY Institutions"] = pd.to_numeric(df["Total Attending SUNY Institutions"]
                                          .replace("<4",4))
```

### Standardizing City Names

City names were standardized by correcting abbreviations, fixing spelling inconsistencies, expanding truncated names, and applying consistent formatting. This ensures that the same city is represented uniformly throughout the dataset, improving the accuracy of grouping and analysis.

```python
duzeltmeler = {
    'St ': 'Saint ',
    'St. ': 'Saint ',
    'Mt ': 'Mount ',
    'Prt Washington': 'Port Washington',
    'Hastings On Huds': 'Hastings On Hudson',
    'Hastings On Hud': 'Hastings On Hudson',
    'Hopewell Junctio': 'Hopewell Junction',
    'Hopewell Jct': 'Hopewell Junction',
    'Springfield Gdns': 'Springfield Gardens',
    'Springfield Gard': 'Springfield Gardens',
    'N Syracuse': 'North Syracuse',
    'S Ozone Park': 'South Ozone Park'
}

# Hatalı isimleri sözlükteki doğrularıyla değiştirme
df['City'] = df['City'].replace(duzeltmeler, regex=False)
```

```python
# String'e çevir ve temel temizlik
df["City"] = (
    df["City"]
    .astype("object")
    .str.strip()
    .replace(r'^\s*$', pd.NA, regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

# Bilinen standartlaştırmalar
city_map = {
    "N White Plains": "North White Plains",
    "Huntington Sta": "Huntington Station",

    # New York'ta resmi kullanımlar
    "Cornwall On Hudson": "Cornwall-on-Hudson",
    "Croton On Hudson": "Croton-on-Hudson",
    "Hastings On Hudson": "Hastings-on-Hudson",

    # Sık yazımlar
    "St Johnsville": "Saint Johnsville",
    "St Regis Falls": "Saint Regis Falls",
    "St James": "Saint James",
    "Ft Edward": "Fort Edward",
    "Ft Plain": "Fort Plain",
    "Mt Vernon": "Mount Vernon",
    "Mt Kisco": "Mount Kisco",
    "Mt Morris": "Mount Morris",
    "Mt Sinai": "Mount Sinai",
    "Mt Upton": "Mount Upton",

    # İstersen bunu kullan
    # "Lake Ronkonkoma": "Ronkonkoma"
}

df["City"] = df["City"].replace(city_map)

# Baştaki/sondaki noktalama işaretlerini temizle
df["City"] = df["City"].str.replace(r"^[,.-]+|[,.-]+$", "", regex=True)

# Boş kalanları NaN yap
df["City"] = df["City"].replace("", pd.NA)
```

```python
# Gruplama
df_plot = (
    df.groupby("Term")["Total Attending SUNY Institutions"]
      .sum()
      .reset_index()
)

# Term'den yılı çıkar
df_plot["Year"] = df_plot["Term"].str.extract(r'(\d{4})').astype(int)

# Yıla göre sırala
df_plot = df_plot.sort_values("Year")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(
    df_plot["Term"],
    df_plot["Total Attending SUNY Institutions"],
    marker="o",
    linewidth=2
)

plt.title("Total Students Attending SUNY Institutions")
plt.xlabel("Term")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```
